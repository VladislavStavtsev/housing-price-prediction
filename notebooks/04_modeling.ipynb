```python
X_clean = df_clean.drop("SalePrice", axis=1)
y_clean = df_clean["SalePrice"]
```

```python
X_fe= df_feature.drop("SalePrice", axis=1)
y_fe = df_feature["SalePrice"]
```

```python
from sklearn.model_selection import train_test_split

Xc_train, Xc_test, yc_train, yc_test = train_test_split(
    X_clean, y_clean, test_size=0.2, random_state=42
)
```

```python
categorical_cols = Xc_train.select_dtypes(include='object').columns
numerical_cols = Xc_train.select_dtypes(exclude='object').columns
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
```

```python
Xc_train_cat = encoder.fit_transform(Xc_train[categorical_cols])
Xc_test_cat = encoder.transform(Xc_test[categorical_cols])

Xc_train_num = Xc_train[numerical_cols].to_numpy()
Xc_test_num = Xc_test[numerical_cols].to_numpy()

print(Xc_train_num.shape)
print(Xc_train_cat.shape)
print(Xc_test_num.shape)
print(Xc_test_cat.shape)
```

```python
Xc_train_final = np.hstack([Xc_train_num, Xc_train_cat])
Xc_test_final = np.hstack([Xc_test_num, Xc_test_cat])
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(Xc_train_final, yc_train)
```

```python
yc_pred = model.predict(Xc_test_final)
```

```python
from sklearn.metrics import mean_squared_error

rmse_clean = np.sqrt(mean_squared_error(yc_test, yc_pred))
print("RMSE clean: ", rmse_clean)
```
