```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("C:/Users/Владислав/housing-price-prediction/data/processed/train_cleaned.csv")
df.info()
```

```python
# Общая площадь подвала
df['TotalBsmtSF'] = df['BsmtFinSF1'] + df['BsmtFinSF2'] + df['BsmtUnfSF']
```

```python
# Площадь законченной части
df['TotalFinishedBsmt'] = df['BsmtFinSF1'] + df['BsmtFinSF2']
df['FinishedBsmtRatio'] = np.where(
    df['TotalBsmtSF'] == 0,
    0,
    df['TotalFinishedBsmt'] / df['TotalBsmtSF']
)
df['FinishedBsmtRatio']
```

```python
# Общее количество ванных комнат
df['TotalBathrooms'] = (
    df['BsmtFullBath'] 
    + df['FullBath'] 
    + 0.5 * df['BsmtHalfBath'] 
    + df['HalfBath']
)
```

```python
# Общая площадь крыльца
df['TotalPorchSF'] = (
    df['EnclosedPorch']
    + df['3SsnPorch']
    + df['ScreenPorch']
)
```

```python
# Вывод новых объясняющих переменных
df[['TotalBathrooms', 'FinishedBsmtRatio', 'TotalBsmtSF', 'TotalPorchSF']].head()
```

```python
# Возраст дома
df['HouseAge'] = df['YrSold'] - df['YearBuilt']

# Возраст с момента реконструкий/добавлений
df['RemodelAge'] = df['YrSold'] - df['YearRemodAdd']

# Общая площадь дома
df['TotalHouseSF'] = df['TotalBsmtSF'] + df['GrLivArea']

# Общая площадь крыльца
df['TotalHouseSF'] = (
    df['OpenPorchSF']
    + df['EnclosedPorch']
    + df['3SsnPorch']
    + df['ScreenPorch']
)

# Наличие второго этажа
df['Has2edFloor'] = (df['2ndFlrSF'] > 0).astype(int)

# Наличие гаража
df['HasGarage'] = (df['GarageArea'] > 0).astype(int)
```

```python
df[['HouseAge', 'RemodelAge', 'TotalHouseSF', 'TotalHouseSF', 'Has2edFloor', 'HasGarage','TotalBathrooms', 'FinishedBsmtRatio', 'TotalBsmtSF', 'TotalPorchSF']].head()
```
